```python
import numpy as np
import pandas as pd
```

```python
df_label = pd.read_csv('qaplib_labels.txt', sep=' ', header=None, skiprows=1,
                 names=['instance', 'obj_label'])
# sort by instance name
df_label = df_label.sort_values(by="instance")
df_label.head()
```

```python
# read txt
seeds = [0, 42, 47, 1111, 1234]
file_name = "results/qaplib_{}.txt"

file_names = [file_name.format(seed) for seed in seeds]
file_names
```

```python
data = []
for file_name in file_names:
    df = pd.read_csv(file_name, sep=" ", header=None, skiprows=1)
    df.columns = ["instance", "time1", "time", "incumbent"]
    df = df[["instance", "time", "incumbent"]]
    # sort by instance name
    df = df.sort_values(by="instance")
    df['gap'] = (df["incumbent"] - df_label["obj_label"]) / df_label["obj_label"] * 100
    data.append(df)
    
df = data[0]
for d in data[1:]:
    df["time"] += d["time"]
    df["incumbent"] += d["incumbent"]
    df["gap"] += d["gap"]
df["time"] /= len(data)
df["incumbent"] /= len(data)
df["gap"] /= len(data)
```

```python
df
```

```python
# groupy by instance with the first 3 letters the same, then average

df_tmp = df[['instance', 'time', 'gap']]
df_tmp['group'] = df['instance'].str[:3]
grouped = df_tmp.groupby('group').agg({
    'time': 'mean',
    'gap': 'mean'
}).reset_index()


# round time to 1 decimal and gap to 2 decimals in percent
grouped['time'] = grouped['time'].round(2)
grouped['gap'] = grouped['gap'].round(1)
grouped['gap'] = grouped['gap'].astype(str) + '\%'


# save to latex table
# grouped.to_latex('qaplib_results_table.tex', index=False, float_format="%.1f")
grouped.to_csv(f'qaplibs_results_table.txt', index=False, float_format="%.2f")
```
