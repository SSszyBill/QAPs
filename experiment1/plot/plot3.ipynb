```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

csv_file = 'result_trace_20260119_152740.csv'
df = pd.read_csv(csv_file)

sns.set_theme(style="whitegrid", context="paper", font_scale=1.2)  # 设置主题风格
palette = {'PGD': '#4c72b0', 'SK': '#c44e52'}  # 定义颜色调色板
g = sns.relplot(
    data=df,
    x = 'Time',
    y = 'Gap',
    hue ='Method',
    style ='Method',
    col = 'Instance',
    palette=palette,
    kind='line',
    height=4,
    aspect=1.2,
    errorbar = ('sd', 1),
    linewidth=2,
    facet_kws={'sharey': False, 'sharex': False},
    col_wrap = 3
)

for ax in g.axes.flatten():
    ax.set_xscale('log')
    ax.set_ylim(-2, 55)
    ax.grid(True, which="both", ls="--", linewidth=0.3)
    ax.axhline(y=0, color='green', linestyle='--', alpha=0.3, linewidth=1)

g.fig.subplots_adjust(top=0.9)
g.fig.suptitle('Convergence Curves on Different QAP Instances', fontsize=16, fontweight='bold')
g.set_axis_labels('Time (s)', 'Gap (%)')

g.savefig('fig0_convergence_curves.png', dpi=300)
plt.show()
```
